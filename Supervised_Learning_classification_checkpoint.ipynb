```python
import pandas as pd

df = pd.read_csv('/content/African_crises_dataset.csv')
df.head()
```

```python
#general information of the dataset
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
#show the summary statistics of the dataset
df.describe(include= 'all')
```

```python
df.head()
```

```python
#encode bank crisis to numerical values
mapping = {'crisis' : 1, 'no_crisis' : 0 }

df['banking_crisis'] = df['banking_crisis'].map(mapping)

df['banking_crisis'].head()
```

```python
#encode country_code using label encoder
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['country_code'] = le.fit_transform(df['country_code'])
```

```python
#handle outliers
Q1 = df['inflation_annual_cpi'].quantile(0.25)
Q3 = df['inflation_annual_cpi'].quantile(0.75)
IQR = Q3 - Q1
df = df[(df['inflation_annual_cpi'] >= Q1 - 1.5*IQR) & (df['inflation_annual_cpi'] <= Q3 + 1.5*IQR)]
```

```python
#create a profile report
from ydata_profiling import ProfileReport

report = ProfileReport(df, title = 'Country_crises', explorative= True)

report.to_file('Country_crisis.html')
```

**Classification models**

*i. Logistic Regression*

```python
#import the libraries
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
df.head()
```

```python
#prepare the data
x = df.drop(['country', 'country_number'], axis=1)
y = df['systemic_crisis']
```

```python
#split the data for train/test
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size= 0.2, random_state= 42)
```

```python
#create a logistic regression model
log_reg_model = LogisticRegression()
log_reg_model.fit(x_train, y_train)
```

```python
#make prediction
training_predictions = log_reg_model.predict(x_train)
testing_predictions = log_reg_model.predict(x_test)

print('Training predictions:', training_predictions)
print()
print('Testing predictions:', testing_predictions)
```

```python
#evaluation of model performance accuracy for training
print('Model training accuracy', log_reg_model.score(x_train, y_train))
```

```python
#evaluation of model performance during testing
print('Model testing accuracy', log_reg_model.score(x_test, y_test))
```

```python
#print the classification report
print(classification_report(y_test, testing_predictions))
```

*ii. KNN*

```python
from sklearn.neighbors import KNeighborsClassifier

knn_model = KNeighborsClassifier(n_neighbors= 4)
knn_model.fit(x_train, y_train)
```

```python
#make some predictions
testing_predictions = knn_model.predict(x_test)
training_predictions = knn_model.predict(x_train)

print('Testing predictions:', testing_predictions)
print()
print('Training predictions:', training_predictions)
```

```python
#evaluate the model
print('Training accuracy', knn_model.score(x_train, y_train))
print()
print('Testing accuracy', knn_model.score(x_test, y_test))
```

```python
#print the classification report
print(classification_report(y_test, testing_predictions))
```

*iii. Decision tree*

```python
#split the data for train/test
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size= 0.2, random_state= 42)
```

```python
#create a decision tree model
dec_tree_model = DecisionTreeClassifier(random_state= 42)
dec_tree_model.fit(x_train, y_train)
```

```python
#Make some predictions
training_predictions = dec_tree_model.predict(x_train)
testing_predictions = dec_tree_model.predict(x_test)

print('Training predictions:', training_predictions)
print()
print('Testing predictions:', testing_predictions)
```

```python
#training evaluation score
print('Training accuracy score:', dec_tree_model.score(x_train, y_train))

#testing evaluation
print('Testing accuracy score:', dec_tree_model.score(x_test, y_test))
```

```python
#print the classification report
print(classification_report(y_test, testing_predictions))
```

*iv. Random Forest*

```python
from sklearn.ensemble import RandomForestClassifier

rand_forest_model = RandomForestClassifier(n_estimators=100, random_state=42)
rand_forest_model.fit(x_train, y_train)
```

```python
#make predictions
training_predictions = rand_forest_model.predict(x_train)
testing_predictions = rand_forest_model.predict(x_test)

print('Training predictions:', training_predictions)
print()
print('Testing predictions:', testing_predictions)
```

```python
#model evaluation
print('Training accuracy score:', rand_forest_model.score(x_train, y_train))
print()
print('Testing accuracy score:', rand_forest_model.score(x_test, y_test))
```

```python
#print classification report
print(classification_report(y_test, testing_predictions))
```
